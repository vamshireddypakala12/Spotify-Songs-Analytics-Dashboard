# 10 Questions to Explore

1. Which artists have the highest average popularity?
2. How has average song popularity changed over time?
3. Are more danceable songs generally more popular?
4. Which years produced the most popular songs?
5. Do energetic songs tend to score higher on popularity?
6. Which songs are the most popular in the dataset?
7. How does explicit content relate to popularity?
8. Are acoustic songs more or less popular than energetic ones?
9. Which audio features have the strongest relationship with popularity?
10. What patterns stand out when comparing tempo, valence, and popularity?

```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from plotly.subplots import make_subplots
```

```python
import pandas as pd

# Load the Spotify songs dataset

df = pd.read_csv("data/spotify_songs.csv")
```

```python
# Explore the dataset
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:\n", df.isna().sum())

# Basic summary statistics
summary = df[["popularity", "danceability", "energy", "tempo", "valence"]].describe()
summary

# Average popularity by year
yearly_popularity = (
    df.groupby("year", as_index=False)["popularity"]
    .mean()
    .sort_values("year")
)

fig1 = px.line(
    yearly_popularity,
    x="year",
    y="popularity",
    markers=True,
    title="Average Popularity by Release Year"
)
fig1.update_layout(template="plotly_white", title_x=0.5)
fig1

# Danceability vs popularity
fig2 = px.scatter(
    df,
    x="danceability",
    y="popularity",
    color="energy",
    hover_data=["track_name", "artist_names"],
    title="Danceability vs Popularity"
)
fig2.update_layout(template="plotly_white")
fig2

# Top artists by average popularity
artist_summary = (
    df.groupby("artist_names")
    .agg(Average_Popularity=("popularity", "mean"), Songs=("track_name", "count"))
    .reset_index()
)
artist_summary = artist_summary[artist_summary["Songs"] >= 5]
artist_summary = artist_summary.sort_values("Average_Popularity", ascending=False).head(10)

fig3 = px.bar(
    artist_summary,
    x="Average_Popularity",
    y="artist_names",
    orientation="h",
    title="Top Artists by Average Popularity"
)
fig3.update_layout(template="plotly_white")
fig3
```

```python
# Answers to the 10 questions using the full dataset

# Q1: Artists with the highest average popularity
artist_popularity = (
    df.groupby("artist_names")
    .agg(Average_Popularity=("popularity", "mean"), Songs=("track_name", "count"))
    .reset_index()
)
artist_popularity = artist_popularity[artist_popularity["Songs"] >= 5]
artist_popularity = artist_popularity.sort_values("Average_Popularity", ascending=False).head(10)
fig1 = px.bar(
    artist_popularity,
    x="Average_Popularity",
    y="artist_names",
    orientation="h",
    title="Top Artists by Average Popularity",
)
fig1.update_layout(template="plotly_white", yaxis_title="Artist", xaxis_title="Average Popularity")
fig1

# Q2: How popularity has changed over time
yearly_popularity = (
    df.groupby("year", as_index=False)["popularity"]
    .mean()
    .sort_values("year")
)
fig2 = px.line(
    yearly_popularity,
    x="year",
    y="popularity",
    markers=True,
    title="Average Popularity by Release Year",
)
fig2.update_layout(template="plotly_white", title_x=0.5, xaxis_title="Year", yaxis_title="Popularity")
fig2

# Q3: Danceability vs popularity
corr_dance = df[["danceability", "popularity"]].corr().loc["danceability", "popularity"]
print(f"Danceability–Popularity correlation: {corr_dance:.3f}")
fig3 = px.scatter(
    df,
    x="danceability",
    y="popularity",
    color="energy",
    hover_data=["track_name", "artist_names"],
    title="Danceability vs Popularity",
)
fig3.update_layout(template="plotly_white", xaxis_title="Danceability", yaxis_title="Popularity")
fig3

# Q4: Top 10 most popular songs
top_songs = df.sort_values("popularity", ascending=False).head(10)[["track_name", "artist_names", "popularity"]]
top_songs

# Q5: Explicit vs non-explicit popularity
explicit_summary = df.groupby("explicit")["popularity"].mean().reset_index(name="Average Popularity")
explicit_summary

# Q6: Years with the highest average popularity
top_years = (
    df.groupby("year", as_index=False)["popularity"]
    .mean()
    .sort_values("popularity", ascending=False)
    .head(10)
)
top_years

# Q7: Energy vs popularity
corr_energy = df[["energy", "popularity"]].corr().loc["energy", "popularity"]
print(f"Energy–Popularity correlation: {corr_energy:.3f}")

# Q8: Acousticness vs energy relation to popularity
acoustic_corr = df[["acousticness", "popularity"]].corr().loc["acousticness", "popularity"]
print(f"Acousticness–Popularity correlation: {acoustic_corr:.3f}")

# Q9: Feature correlations with popularity
feature_corr = (
    df[["danceability", "energy", "valence", "acousticness", "speechiness", "liveness", "tempo", "popularity"]]
    .corr()["popularity"]
    .drop("popularity")
    .sort_values(ascending=False)
    .reset_index()
)
feature_corr.columns = ["Feature", "Correlation with Popularity"]
feature_corr

# Q10: Tempo, valence, and popularity
fig10 = px.scatter(
    df,
    x="tempo",
    y="valence",
    size="popularity",
    color="popularity",
    hover_data=["track_name", "artist_names"],
    title="Tempo vs Valence (size = popularity)",
)
fig10.update_layout(template="plotly_white", xaxis_title="Tempo", yaxis_title="Valence")
fig10
```
